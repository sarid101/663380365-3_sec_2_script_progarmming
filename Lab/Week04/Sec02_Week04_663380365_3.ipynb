# lab4_1.py - Student Management System (List Operations)

````python
# lab4_1_challenge.py - Student Management System with Menu
print("--- Lab 4.1: Student Management System (Challenge Version) ---")

student_names = []

while True:
    print("\n===== STUDENT MANAGEMENT SYSTEM =====")
    print("1. Add Students")
    print("2. Find a Student")
    print("3. Remove a Student")
    print("4. Sort Students")
    print("5. Count & View All Students")
    print("6. Exit")

    choice = input("Enter your choice (1-6): ")

    if choice == '1':
        print("\n[1] Add Students")
        count = int(input("How many students do you want to add? "))
        for i in range(count):
            name = input(f"Enter student name #{i + 1}: ")
            student_names.append(name)
        print("Students added successfully!")

    elif choice == '2':
        print("\n[2] Find a Student")
        search_name = input("Enter a name to search for: ")
        if search_name in student_names:
            position = student_names.index(search_name)
            print(f"Found '{search_name}' at index {position}")
        else:
            print(f"'{search_name}' not found in the list")

    elif choice == '3':
        print("\n[3] Remove a Student")
        remove_name = input("Enter a name to remove: ")
        if remove_name in student_names:
            student_names.remove(remove_name)
            print(f"'{remove_name}' has been removed")
        else:
            print(f"'{remove_name}' not found, nothing removed")

    elif choice == '4':
        print("\n[4] Sort Students")
        student_names.sort()
        print(f"Sorted list: {student_names}")

    elif choice == '5':
        print("\n[5] Count & View Students")
        print(f"Current Student List: {student_names}")
        print(f"Total students: {len(student_names)}")

    elif choice == '6':
        print("\nThank you for using the system. Goodbye!")
        break
    else:
        print("Invalid choice! Please select between 1-6.")
````

## 3. ส่วนการออกจากโปรแกรมและการดักจับข้อผิดพลาด (Exit & Error Handling)

```python
    elif choice == '6':

```

* **`choice == '6'`**: ถ้าผู้ใช้พิมพ์เลข 6 โปรแกรมจะสั่ง **`break`** เพื่อหยุดการทำงานของลูป `while True` ทันที ทำให้โปรแกรมจบการทำงานอย่างสมบูรณ์
* **`else:`**: หากผู้ใช้พิมพ์สิ่งที่นอกเหนือจาก '1' ถึง '6' (เช่น พิมพ์ข้อความหรือตัวเลขอื่น) โปรแกรมจะแจ้งเตือนว่าข้อมูลไม่ถูกต้อง และวนกลับไปแสดงเมนูใหม่อีกครั้งอย่างปลอดภัยครับ


````python
# lab4_2_challenge.py - Geometric Calculations with User Input
import math

print("--- Lab 4.2: Geometric Calculations (Challenge Version) ---")

print("\n--- Enter Coordinates for Point 1 ---")
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
user_point1 = (x1, y1) # ประกอบร่างเป็น Tuple

print("\n--- Enter Coordinates for Point 2 ---")
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))
user_point2 = (x2, y2) # ประกอบร่างเป็น Tuple

# แสดงผลพิกัดที่ผู้ใช้กรอก
print(f"\nYour Point 1: {user_point1}")
print(f"Your Point 2: {user_point2}")

# คำนวณระยะทาง
user_distance = math.sqrt((user_point2[0] - user_point1[0]) ** 2 + (user_point2[1] - user_point1[1]) ** 2)
print(f"Calculated Distance: {user_distance:.4f}") # แสดงทศนิยม 4 ตำแหน่ง
````
