Activity 1.2: "Python Puzzlers" - Identifying Data Types & Output
Format: Individual or pair work, followed by a class review.
Instructions: Provide a set of small Python code snippets (on a handout or projected). For each snippet, students must:

```python
# Puzzler 1
x = 10
y = 3.5
z = x + y
print(z)
print(type(z))
```

Output: 13.5
<class 'float'>

คำอธิบาย: ข้อนี้แสดงให้เห็นถึง การแปลงชนิดข้อมูลโดยอัตโนมัติ (Implicit Type Conversion) ใน Python เมื่อมีการนำตัวเลขจำนวนเต็ม (int เช่นค่าของ x) มาดำเนินการทางคณิตศาสตร์ร่วมกับตัวเลขทศนิยม (float เช่นค่าของ y) Python จะทำการเลื่อนขั้น (promote) ชนิดข้อมูลของจำนวนเต็มให้กลายเป็นทศนิยมโดยอัตโนมัติ (มอง 10 เป็น 10.0) เพื่อป้องกันส่วนของทศนิยมสูญหาย ผลลัพธ์สุดท้ายที่ได้จึงมีชนิดข้อมูลเป็น float เสมอ

```python
# Puzzler 2
name = "Risa"
age = "21" # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))
```

Output: Risa is 21 years old.
<class 'str'>

คำอธิบาย: แม้ว่า 21 จะเป็นตัวเลข แต่เนื่องจากถูกครอบด้วยเครื่องหมายคำพูด ("") ตัวแปร age จึงมีชนิดข้อมูลเป็นสตริง (ข้อความ หรือ str) ดังนั้นตัวดำเนินการ + ในที่นี้จึงทำหน้าที่เชื่อมต่อข้อความ (Concatenation) ทั้งหมดเข้าด้วยกันเป็นประโยคเดียว ไม่ได้เป็นการบวกเลขทางคณิตศาสตร์ และเมื่อตรวจสอบชนิดข้อมูลของตัวแปร age ด้วยฟังก์ชัน type() จึงได้ผลลัพธ์ออกมาเป็น <class 'str'>

```python
# Puzzler 3
is_sunny = True
is_raining = False
print(is_sunny and is_raining)
print(type(is_sunny))
```

Output: False
<class 'bool'>

คำอธิบาย: ข้อนี้ทดสอบความเข้าใจเกี่ยวกับ ชนิดข้อมูลบูลีน (Boolean) และ ตัวดำเนินการทางตรรกศาสตร์ (Logical Operators) ค่า True (จริง) และ False (เท็จ) จัดเป็นชนิดข้อมูลประเภท bool ใน Python เมื่อเราใช้ตัวดำเนินการ and (และ) ผลลัพธ์จะออกมาเป็น True ก็ต่อเมื่อค่าทั้งสองข้างเป็นจริงเท่านั้น ในกรณีนี้มีค่าหนึ่งเป็น False ผลลัพธ์ของการประเมิน is_sunny and is_raining จึงออกมาเป็น False และเมื่อตรวจสอบชนิดข้อมูลด้วย type() ก็จะได้รับการยืนยันว่าเป็นคลาส bool

```python
# Puzzler 4
# What happens here? - TypeError: cannot add str and int directly
# num_str = input("Enter a number: ")
# result = num_str + 5  # This would cause TypeError
# print(result)

# Fixed version: convert input to int before adding
num_str = input("Enter a number: ")
result = int(num_str) + 5  # Convert str to int first
print(result)
```

Output:
Error (เกิดข้อผิดพลาดประเภท TypeError: can only concatenate str (not "int") to str) หลังจากที่เราพิมพ์ข้อความหรือตัวเลขลงไปแล้วกด Enter

Data Type: ตัวแปร num_str มีประเภทข้อมูลเป็น str (String)
ตัวเลข 5 มีประเภทข้อมูลเป็น int (Integer)

Reason: ฟังก์ชัน input() ในภาษา Python มีคุณสมบัติพิเศษคือ จะส่งกลับค่าเป็นข้อความ (str) เสมอ ไม่ว่าผู้ใช้จะพิมพ์ตัวเลขใด ๆ ลงไปก็ตาม (เช่น ถ้าเราพิมพ์เลข 10 ตัวแปร num_str จะเก็บค่าเป็นข้อความ "10")

ในบรรทัดต่อมา โค้ดพยายามนำ num_str (ซึ่งเป็นข้อความ) ไปใช้เครื่องหมาย + กับเลข 5 (ซึ่งเป็นจำนวนเต็ม) ซึ่ง Python ไม่อนุญาตให้เอาข้อความมาบวกคำนวณกับตัวเลขตรง ๆ จึงส่งผลให้ระบบหยุดทำงานและแจ้งเตือน TypeError

# แก้ไขโดยแปลง input ให้เป็นจำนวนเต็ม (int) ก่อนนำไปบวก
num_str = input("Enter a number: ")
result = int(num_str) + 5  # แปลงข้อความให้เป็นตัวเลขก่อนบวก
print(result)
