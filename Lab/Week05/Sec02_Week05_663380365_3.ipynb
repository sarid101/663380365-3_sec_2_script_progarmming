## Lab 5.1: Simple Contact Book Manager

```python
# lab5_1.py
# Lab 5.1: Simple Contact Book Manager (Dictionary Focus)

contacts = {}


def normalize(name):
    """Challenge: ทำให้ชื่อไม่สนตัวพิมพ์เล็ก/ใหญ่ โดยแปลงเป็น .title() เสมอ"""
    return name.strip().title()


def add_or_update_contact():
    name = normalize(input("Enter contact name: "))
    if not name:
        print("Name cannot be empty.")
        return

    phone = input("Enter phone number: ").strip()
    email = input("Enter email: ").strip()

    if name in contacts:
        contacts[name] = {"phone": phone, "email": email}
        print(f"Contact '{name}' updated.")
    else:
        contacts[name] = {"phone": phone, "email": email}
        print(f"Contact '{name}' added.")


def view_contact():
    name = normalize(input("Enter contact name to view: "))
    details = contacts.get(name)
    if details:
        print(f"Name : {name}")
        print(f"Phone: {details['phone']}")
        print(f"Email: {details['email']}")
    else:
        print("Contact not found.")


def list_all_contacts():
    if not contacts:
        print("Contact book is empty.")
        return
    print("--- All Contacts ---")
    for name, details in contacts.items():
        print(f"{name} | Phone: {details['phone']} | Email: {details['email']}")


def delete_contact():
    name = normalize(input("Enter contact name to delete: "))
    removed = contacts.pop(name, None)
    if removed is not None:
        print(f"Contact '{name}' deleted.")
    else:
        print("Contact not found.")


def main():
    while True:
        print("\n===== Contact Book =====")
        print("1. Add/Update Contact")
        print("2. View Contact Details")
        print("3. List All Contacts")
        print("4. Delete Contact")
        print("5. Exit")
        choice = input("Choose an option (1-5): ").strip()

        if choice == "1":
            add_or_update_contact()
        elif choice == "2":
            view_contact()
        elif choice == "3":
            list_all_contacts()
        elif choice == "4":
            delete_contact()
        elif choice == "5":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1-5.")


main()
```

## Lab 5.2: Course Enrollment & Collaboration

```python
# lab5_2.py
# Lab 5.2: Course Enrollment & Collaboration (Set Focus)

python_students = {"S001", "S003", "S005", "S007", "S009"}
java_students = {"S002", "S003", "S006", "S007", "S010"}

print(f"Python students: {sorted(python_students)}")
print(f"Java students  : {sorted(java_students)}")

# 1. Students in both courses (intersection)
both = python_students & java_students
print(f"\nStudents in both courses (&): {sorted(both)}")
print(f"Using intersection()        : {sorted(python_students.intersection(java_students))}")

# 2. All unique students (union)
all_unique = python_students | java_students
print(f"\nAll unique students (|): {sorted(all_unique)}")
print(f"Using union()          : {sorted(python_students.union(java_students))}")

# 3. Students only in Python (difference)
only_python = python_students - java_students
print(f"\nStudents only in Python (-): {sorted(only_python)}")
print(f"Using difference()         : {sorted(python_students.difference(java_students))}")

# 4. Students in only one course (symmetric difference)
exclusive = python_students ^ java_students
print(f"\nStudents in only one course (^): {sorted(exclusive)}")
print(f"Using symmetric_difference()   : {sorted(python_students.symmetric_difference(java_students))}")

# 5. Remove duplicates from a list
all_students_list = ["S001", "S003", "S005", "S001", "S002"]
unique_students_list = list(set(all_students_list))
print(f"\nOriginal list: {all_students_list}")
print(f"Unique list  : {sorted(unique_students_list)}")
```

## Activity 5.2: Word Frequency Counter

```python
# activity5_2.py
# Activity 5.2: Word Frequency Counter (using Dictionary)

sentence = input("Enter a sentence: ")
sentence = sentence.lower()
words = sentence.split()

word_counts = {}

for word in words:
    # Challenge: remove punctuation around the word
    word = word.strip('.,!?"\'')
    if not word:
        continue
    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print(word_counts)
```
