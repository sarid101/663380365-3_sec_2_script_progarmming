```python
# นายสาริษฐ์ บุตรช่วง 663380365-3 sec 02
# --- Lab 3.1 Part 1: Single Number Multiplication Table ---
n = int(input("Enter a number for its multiplication table (e.g., 7): "))
for i in range(1, 13):
    print(f"{n} * {i} = {n * i}") #คำใบ้ print ออกมาในรูปแบบ {a} * {b} = {a*b}


```

```python
# --- Lab 3.1 Part 2: Full 12x12 Multiplication Grid ---
print("12x12 Multiplication Table:")
for i in range(1, 13): #เติมตัวแปรให้สมบูรณ์
    for j in range(1, 13): #เติมตัวแปรให้สมบูรณ์
        print(f"{i * j}", end="\t") #คำใบ้ print ออกมาในรูปแบบ ตาราง เหมือนกันกับด้านบน แต่ต้องมีการเว้นวรรคให้เป็นตาราง
    print()

print()  # spacer
```

```python
# --- Lab 3.2 Part 1: Countdown Timer ---
start = int(input("Enter starting number for countdown: "))
while start >= 0:
    print(start)
    start -= 1
print("Blast off!")#คำใบ้ print ออกมาในรูปแบบ Countdown: {start} และลดค่าของ start ลง 1
```

```python
# --- Lab 3.2 Part 2: Simple Guessing Game ---
secret_number = 42

while True:
    guess = int(input("Guess the number: "))

    if guess < secret_number:  # ถ้าตัวเลขที่ทาย น้อยกว่า ตัวเลขที่ซ่อนอยู่
        print(f"{guess} is too low! Try again.")
    elif guess > secret_number:  # ถ้าตัวเลขที่ทาย มากกว่า ตัวเลขที่ซ่อนอยู่
        print(f"{guess} is too high! Try again.")
    else:  # ถ้าทายถูก (เท่ากับ secret_number)
        print("Congratulations! You guessed it!")
        break
```
