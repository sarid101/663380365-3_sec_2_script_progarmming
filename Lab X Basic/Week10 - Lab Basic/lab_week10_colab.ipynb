## 4. ผลลัพธ์ใน output_sales_report.xlsx

```python
wb = openpyxl.load_workbook('data/output_sales_report.xlsx')
ws = wb.active
print('Sheet:', ws.title)
for row in ws.iter_rows(values_only=True):
    print(row)
print('Merged cells:', ws.merged_cells.ranges)
```

## 5. แสดงเป็นตาราง

```python
import pandas as pd
pd.read_excel('data/output_sales_report.xlsx')
```

## 6. ตรวจรูปแบบเซลล์ (formatting)

```python
for ref in ['A1', 'D2', 'A7', 'D7']:
    c = ws[ref]
    print(f"{ref:3} value={c.value!r:16} bold={c.font.b!s:5} fill={c.fill.start_color.rgb}  format={c.number_format}")
```
