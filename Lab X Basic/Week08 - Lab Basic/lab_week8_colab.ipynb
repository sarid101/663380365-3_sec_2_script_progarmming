## 5. ผลลัพธ์ JSON (อัปเดต stock + เพิ่ม PROD004)

```python
import json
with open('data/output_updated_inventory.json', encoding='utf-8') as f:
    inventory = json.load(f)
for p in inventory:
    print(f"{p['id']:8} {p['name']:22} stock={p['stock']:4}  brand={p['details']['brand']}")
```
