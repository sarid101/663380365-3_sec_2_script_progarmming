## 3. ข้อความที่ดึงจาก input_article.pdf

```python
from PyPDF2 import PdfReader
reader = PdfReader('documents/input_article.pdf')
for i, page in enumerate(reader.pages, 1):
    print(f'===== Page {i} =====')
    print(page.extract_text())
```

## 4. ตรวจ merged_output.pdf

```python
merged = PdfReader('documents/merged_output.pdf')
print('จำนวนหน้า:', len(merged.pages))
for i, page in enumerate(merged.pages, 1):
    print(i, page.extract_text())
```

## 5. อ่านเนื้อหา output_report.docx

```python
import docx
doc = docx.Document('documents/output_report.docx')
for p in doc.paragraphs:
    if p.text.strip():
        print(f'[{p.style.name}] {p.text[:80]}')
print('\n--- Table ---')
for row in doc.tables[0].rows:
    print([c.text for c in row.cells])
```
